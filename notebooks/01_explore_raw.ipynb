```python
import glob, os, json
import pandas as pd

base = r"C:\Users\bisar\Downloads\hearthwell_data_kit"

frames = []
for f in sorted(glob.glob(base + r"\reviews\*.json")):
    with open(f, encoding="utf-8") as fh:
        df = pd.DataFrame(json.load(fh))
    df["source_file"] = os.path.basename(f)
    frames.append(df)
reviews = pd.concat(frames, ignore_index=True)

frames = []
for f in sorted(glob.glob(base + r"\tickets\*.csv")):
    df = pd.read_csv(f, dtype=str)
    df["source_file"] = os.path.basename(f)
    frames.append(df)
tickets = pd.concat(frames, ignore_index=True)

products = pd.read_csv(base + r"\products.csv", dtype=str)
customers = pd.read_csv(base + r"\customers.csv", dtype=str)
```

```python
# ---- Q1: how big is the data, and is any file different? ----
print("Reviews:", len(reviews), "rows | Tickets:", len(tickets), "rows")
print("\nRows per review file:")
print(reviews.groupby("source_file").size().describe())
print("\nRows per ticket file:")
print(tickets.groupby("source_file").size())

print("\nFirst 3 column names of each ticket file:")
for f in sorted(glob.glob(base + r"\tickets\*.csv")):
    print(os.path.basename(f), list(pd.read_csv(f, nrows=0).columns)[:3])

print("\nTypes found in the rating column:")
print(reviews["rating"].map(type).value_counts())
print("Files where rating is text:", reviews[reviews["rating"].map(type) == str]["source_file"].unique())

# ---- Q2: duplicates ----
n = len(reviews)
dup_id  = reviews["review_id"].duplicated().sum()
dup_row = reviews.drop(columns="source_file").duplicated().sum()
print(f"\nReviews: repeated review_id = {dup_id} ({dup_id/n*100:.1f}%) | fully identical rows = {dup_row}")

t = len(tickets)
tdup_id  = tickets["ticket_id"].duplicated().sum()
tdup_row = tickets.drop(columns="source_file").duplicated().sum()
print(f"Tickets: repeated ticket_id = {tdup_id} ({tdup_id/t*100:.1f}%) | fully identical rows = {tdup_row}")

# ---- Q3: missing values ----
print("\nReviews, % missing per column:")
print((reviews.isna().sum() / len(reviews) * 100).round(1))
print("\nTickets, % missing per column:")
print((tickets.isna().sum() / len(tickets) * 100).round(1))
print("\nReviews with an empty title:", (reviews["title"] == "").sum())
```

```python
# ---- Q4: dates ----
shape = (tickets["created_date"]
         .str.replace(r"\d", "d", regex=True)
         .str.replace(r"[A-Za-z]", "a", regex=True))
print("Ticket date formats (d = digit, a = letter):")
print(shape.value_counts())

slash = tickets[tickets["created_date"].str.contains("/")]["created_date"]
first  = slash.str.split("/").str[0].astype(int)
second = slash.str.split("/").str[1].astype(int)
print("\nSlash dates:", len(slash), "| first number > 12:", (first > 12).sum(), "| second number > 12:", (second > 12).sum())

ts = pd.to_datetime(reviews["created_at"], utc=True)
print("\nReview dates run from", ts.min(), "to", ts.max())

# ---- Q5: text problems ----
html_n = reviews["body"].str.contains(r"<[^>]+>|&\w+;", regex=True).sum()
print(f"\nReviews with HTML: {html_n} ({html_n/len(reviews)*100:.1f}%)")
for label, pat in [("quoted reply ('wrote:')", "wrote:"),
                   ("'Sent from my iPhone'", "Sent from my iPhone"),
                   ("confidentiality notice", "CONFIDENTIALITY NOTICE"),
                   ("'Regards,' sign-off", "Regards,")]:
    c = tickets["description"].str.contains(pat, regex=False).sum()
    print(f"Tickets with {label}: {c} ({c/len(tickets)*100:.1f}%)")

short = reviews["body"].str.len() < 10
print("\nReviews with body under 10 characters:", short.sum())
print(reviews[short]["body"].value_counts().head(10))

# ---- Q6: broken links ----
print("\nReview product_ids not in products:", (~reviews["product_id"].dropna().isin(products["product_id"])).sum())
print("Ticket product_ids not in products:", (~tickets["product_id"].dropna().isin(products["product_id"])).sum())
print("Ticket emails not in customers:", (~tickets["customer_email"].isin(customers["email"])).sum())
```

```python
dups = reviews[reviews["review_id"].duplicated(keep=False)].copy()
as_text = dups.drop(columns="source_file").astype(str).drop_duplicates()
still_different = as_text[as_text["review_id"].duplicated(keep=False)]
print("IDs whose copies still differ when everything is compared as text:", still_different["review_id"].nunique())

if still_different["review_id"].nunique() > 0:
    print(dups[dups["review_id"] == still_different["review_id"].iloc[0]].T)
```
